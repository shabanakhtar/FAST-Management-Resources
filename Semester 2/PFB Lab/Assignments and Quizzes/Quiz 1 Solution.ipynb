```python
def lesser_of_two_evens(a, b):
    # check if both of the numbers are even
    if a % 2 == 0 and b % 2 == 0:
        if a < b:
            return a
        else:
            return b
    else:
        if a > b:
            return a
        else:
            return b

#tests
first = lesser_of_two_evens(10,6) #should be 6
second = lesser_of_two_evens(9,6) #should be 9

print(first)
print(second)
```
