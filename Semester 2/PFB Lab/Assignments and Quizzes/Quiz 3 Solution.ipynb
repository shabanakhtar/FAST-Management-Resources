```python
def makes_twenty(one, two):
    if one == 20:
        return True
    elif two == 20:
        return True
    elif one + two == 20:
        return True
    else:
        return False

#test
first = makes_twenty(10, 10) # 10 + 10 --> 20, so true
second = makes_twenty(20, 5) # 20 mentioned, so true
third = makes_twenty(10, 5) # total not 20, nor 20 mentioned, so false

print(first)
print(second)
print(third)
```
