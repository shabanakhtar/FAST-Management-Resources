```python
def count(sample_string):

    upper_case = 0
    lower_case = 0
    
    for ch in sample_string:
        if ch.isupper():
            upper_case = upper_case + 1
        if ch.islower():
            lower_case = lower_case + 1
    
    print(f"String: {sample_string}")
    print(f"Number of upper case letters: {upper_case}")
    print(f"Number of lower case letters: {lower_case}")
    print()

count("BRead") # should be 2 uppers, 3 lowers
count("BANDit") # should be 4 uppers, 2 lowers
```
