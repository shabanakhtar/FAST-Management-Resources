```python
def check_function(text): # animal_crackers wasn't really a good function name
    words = text.split()   # this splits the string into two parts
    
    first_word = words[0]
    second_word = words[1]
    
    # compares the first letters of each word. Also turn both to lower to avoid problems
    if first_word[0].lower() == second_word[0].lower():
        return True
    else:
        return False

#tests
first = check_function("Brandon bear") # should be true
second = check_function("Shaban Akhtar") # should be false

print(first)
print(second)
```
