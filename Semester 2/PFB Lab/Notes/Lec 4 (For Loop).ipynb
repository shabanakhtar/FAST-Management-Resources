# The for Loop: A Count-Controlled Loop

## A count-controlled loop iterates a specific number of times.

A <code>for</code> loop acts as an iterator in Python; it goes through items that are in a *sequence* or any other iterable item. Objects that we've learned about that we can iterate over include strings, lists, tuples, and even built-in iterables for dictionaries, such as keys or values.

Here's the general format for a <code>for</code> loop in Python:

    for item in object:
        statements to do stuff
    

The variable name used for the item is completely up to the coder, so use your best judgment for choosing a name that makes sense and you will be able to understand when revisiting your code. This item name can then be referenced inside your loop, for example if you wanted to use <code>if</code> statements to perform checks.

Let's go ahead and work through several example of <code>for</code> loops using a variety of data object types. We'll start simple and build more complexity later on.

### The range() Function with the for Loop
**The range() function in a Python for loop is used to perform an action a specific number of times or to iterate over a sequence of numbers. It is particularly useful when you need to access the index of elements or simply count iterations.**

**range() function generates the immutable sequence of numbers starting from the given start integer to the stop integer. The range() is a built-in function that returns a range object that consists series of integer numbers, which we can iterate using a for loop.**

**The range() function is memory-efficient because it generates numbers one by one as needed (lazy evaluation), rather than creating an entire list in memory.**

### Syntax and Parameters

The range() function can be called with one, two, or three arguments, all of which must be integers: 

* range(stop): Generates a sequence of integers <code>starting from 0 up to, but not including, the stop value</code>, with a default step of 1.
* range(start, stop): Generates a sequence from the <code>start value (inclusive) up to, but not including, the stop value, with a default step of 1</code>.
* range(start, stop, step): Generates a sequence from <code>start to stop, incrementing or decrementing by the step value</code>. The step cannot be zero.

## Example 9
let's create a <code>for</code> loop that sums up the list:

```python
# Start sum at zero
list_sum = 0 

for num in list1:
    list_sum = list_sum + num

print(list_sum)
```

# __Input validation loop:__
#### Ask the user for two numbers between 1 and 100. Then count from the lower number to the higher number. Print the result to the screen.

```python
num_1 = int(input("Pleas enter a number between 1-100:> "))
num_2 = int(input("Pleas enter a number between 1-100:> "))

while (num_1 < 0) or (num_2 < 0) or (num_1 > 100) or (num_2 > 100) or (num_1==num_2):
    print("Numbers must be different values between 1 and 100, try again")
    num_1 = int(input("Pleas enter a number between 1-100:> "))
    num_2 = int(input("Pleas enter a number between 1-100:> "))
    
if num_1 < num_2:
    for i in range(num_1, num_2+1):
        print(i, end=' ')
else:
    for i in range(num_2, num_1+1):
        print(i, end=' ')
```

#### Ask the user to input a sequence of numbers. Then calculate the mean and print the result.

```python
user_input = input("Please enter a number:> ")
numbers = []
while user_input.lower() != 'exit':
    while not user_input.isdigit():
        print("That is not a number! Numbers only please:> ")
        user_input = input("Try again:> ")
    numbers.append(int(user_input))
    user_input = input("Please enter next number:> ")

total = 0    
for num in numbers:
    total = total + num

print(f"Mean is = {total / len(numbers):,.2f}")
print("Mean: ", sum(numbers) / len(numbers))
```

## Nested Loop:
A loop that is inside another loop is called a nested loop.

```python
star=5

for row in range(star):    # outer loop run 5 times
    for col in range(row): # inner loop depend on outer loop counter
        print(col, end=" ")
    print()
```

```python
for row in range(4, 0, -1):
    for col in range(row):
        print(col , end=" ")
    print()
```

```python
for row in range(1, 4):  # Outer loop controls rows
    print(row)
    for col in range(1, 4):      # Inner loop controls columns
        print(f'({row},{col})', end='')  # Print coordinates
    print()
```

##### This program averages test scores. 
* It asks the user for 
* the number of students and
* the number of test scores per student.

```python
# Get the number of students
no_students = int(input("How many students do you have? "))

# Get the number of test scores per student
no_test_scores = int(input("How many test scores per student? "))
```

```python
# Determine each student's average test score
for student in range(no_students):
    print(f"Student number {student + 1}")
    print("----------------")
    total=0
    
    # Get the student's test scores
    for test_scores in range(no_test_scores):
        score = int(input(f"Test number {test_scores + 1}: "))
        
        # Add the score
        total = total + score
    average = total / no_test_scores
    print(f"The average for student number {student + 1} is: {average:.1f}")
    print()
```
