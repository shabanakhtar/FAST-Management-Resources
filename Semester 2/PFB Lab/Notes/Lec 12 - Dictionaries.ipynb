# Dictionaries

We've been learning about *sequences* in Python but now we're going to switch gears and learn about *mappings* in Python. If you're familiar with other languages you can think of these Dictionaries as hash tables. 

This section will serve as a brief introduction to dictionaries and consist of:

    1.) Constructing a Dictionary
    2.) Accessing objects from a dictionary
    3.) Nesting Dictionaries
    4.) Basic Dictionary Methods

So what are mappings? Mappings are a collection of objects that are stored by a *key*, unlike a <code>sequence that stored objects by their relative position</code>. This is an important distinction, since mappings won't retain order since <code>they have objects defined by a key.</code>

A Python dictionary consists of a <code>key and then an associated value.</code> That value can be almost any Python object.

Its important to note that dictionaries are very flexible in the data types they can hold. For example:

```python
my_dict = {'key1':123,
           'key2':[12,23,33],
           'key3':['item0','item1','item2']}
```

```python
# Let's call items from the dictionary
my_dict['key3']

# 'key3':['item0','item1','item2']
#          [0]      [1]     [2]
```

```python
# Can call an index on that value
my_dict['key3'][1]
```

```python
# Can then even call methods on that value
# dicttionaryName['keyName'][index].lower()

my_dict['key3'][0].upper()
```

We can affect the values of a key as well. For instance:

```python
my_dict['key1']
```

```python
# Subtract 123 from the value
# x = x  - 123
my_dict['key1'] = my_dict['key1'] - 123
```

```python
#Check
my_dict['key1']
```

A quick note, Python has a built-in method of doing a self subtraction or addition (or multiplication or division). We could have also used += or -= for the above statement. For example:

```python
# Set the object equal to itself minus 123 
my_dict['key1'] += 123
my_dict['key1']

my_dict['key1'] -= 123
my_dict['key1']
```

## Nesting with Dictionaries

Hopefully you're starting to see how powerful Python is with its flexibility of nesting objects and calling methods on them. Let's see a dictionary nested inside a dictionary:

```python
# Dictionary nested inside a dictionary nested inside a dictionary
d = {'key1':{'nestkey':{'subnestkey':'value'}}}
d
```

```python
d['key1']
```

```python
d['key1']['nestkey']
```

Wow! That's a quite the inception of dictionaries! Let's see how we can grab that value:

```python
# Keep calling the keys
d['key1']['nestkey']['subnestkey']
```

## Iteration over keys, values, and items
Dictionaries can be iterated over using the keys(), values() and items() methods. For example:

```python
d = {'k1':10,'k2':20}
d
```

```python
for k in d.keys():
    print(k)
```

```python
for v in d.values():
    print(v)
```

```python
for item in d.items():
    print(item)
```

```python
for keys, values in d.items():
    print(keys, values)
```

# Viewing keys, values and items
