# Repetition Structures

## In Python, repetition structures (or loops) are used to execute a block of code repeatedly. The two main types of loops are the <code>for loop</code> and the <code>while loop</code>, which are used to manage the flow of a program. 

# Condition-Controlled and Count-Controlled Loops
**1) A condition-controlled loop is the <code>while loop</code>. This type of loop repeatedly executes a block of code as long as a specified condition remains <code>True</code>.**

**2) A count-controlled loop is the <code>for loop</code>, typically used with the built-in range() function. A <code>for loop</code> is ideal when you know the <code>exact number of iterations needed in advance</code>.**
#### In Python, you use the <code>*while*</code> *statement to write a* <code>*condition-controlled loop*</code>,
#### and you use the <code>*for*</code> *statement to write a* <code>*count-controlled loop*</code>

# while Loops

The <code>while</code> statement in Python is one of most general ways to perform iteration. A <code>while</code> statement will repeatedly execute a single statement or group of statements as long as the condition is true. The reason it is called a 'loop' is because the code statements are looped through over and over again until the condition is no longer met.

The general format of a while loop is:

    while test condition:
        code statements


    while test condition:
        code statements
    else:
        final code statements

A while loop is used to execute a block of statements repeatedly until a given condition is satisfied. When the condition becomes false, the line immediately after the loop in the program is executed.

**The while loop is known as a <code>pretest loop</code>,**
**Which means it <code>tests</code> its <code>condition before</code> performing an iteration.**
**Because the <code>test</code> is done at the beginning of the loop.**

In below code, loop runs as long as the condition cnt < 3 is true. It increments the counter by 1 on each iteration and prints "Fast School of Management" three times.

```python
cnt = 0
while (cnt < 3):
    print("Fast School of Management")
    cnt = cnt + 1
```

**Each execution of the body of a loop is known as an iteration.**

**Print Numbers from 1-10 using while loop**

```python
num=1
while num<=10:
    print(num)
    num = num + 1
```

**Print numbers 10 -1 using while loop**

```python
num = 10
while num>=1:
    print(num)
    num = num - 1
```

In below code, loop runs as long as the condition cnt < 10 is true. It increments the counter by 1 on each iteration and prints 10 times
* "cnt is currently: 0" 
* "cnt is still less than 10, adding 1 to cnt"

#### Boolean Variables
###### A Boolean variable can reference one of two values: True or False.
###### Boollean variables are commonly used as <code>flags</code>, which indicate whether <code>specific conditions exist</code>.
###### A <code>flag</code> is a variable that signals when some <code>condition exists</code> in the program.
###### When the <code>flag</code> variable is set to <code>flag = True</code> , it means the condition does <code>exist or execute</code>.
###### When the <code>flag</code> variable is set to <code>flag = False</code>, it indicates the condition <code>does not exist or not execute</code>.

**If the condition of a while loop is True before the first iteration, it will execute the loop body at least once, but it will continue to iterate as long as the condition remains True.**

```python
flag= True
while flag:
    print("Condition is True")
    flag=False
```

##### Formatting
##### Formatting Integers
#### Float formatting follows `"The number is: {value:{width}{comma}}"`

```python
number = 123456

print(f'The number is:{number}')
# the number 123456 is printed with no special formatting
print(f'The number is:{number:d}')
```

```python
number = 123456
# the number 123456 is printed 
# with a comma separator
print(f'The number is:{number:,d}')
```

```python
number = 123456
# the number 123456 is printed 
# with a comma separator
# in a field that is 20 spaces wide
print(f'The number is:{number:20,d}')
```

#### Float formatting follows `"The number is: {value:{width}{comma}.{precision}}"`

```python
number = 12345.6789

# display a floating-point value
print(f'The number is: {number}')

# display a floating-point value with comma separators.
print(f'The number is: {number:,}')

# display a floating-point value 
# rounded to 2 decimal places, 
# with comma separators.
print(f'The number is: {number:,.2f}')

# display a floating-point value 
# rounded to 2 decimal places, 
# with comma separators,
# in a field that is 12 spaces wide:
print(f'The number is: {number:12,.2f}')
```

### Write a Python program that asks
* the user to enter <code>positive numbers repeatedly.</code> 
* After each positive number, it should <code>add it to a running total.</code>
* The program should <code>stop</code> asking for <code>input and **print the total,**</code> 
* when the user <code>enters 0 or a negative number.</code>

```python
num=1
total=0
while num>0:
    num = float (input(" Enter positive number or 0 / negitive to stop: "))
    if num > 0:
        total = total + num
        
print("Running Total: ", total)
```

### Implement a while loop 
* that repeatedly prompts the user to guess a secret number (e.g: 7) until they guess correctly.
* Provide feedback such as "Too high", "Too low", or "Guessed Correctly!"

```python
secret_number=7
guess= int(input("Guess the secret number: "))
while guess != secret_number:
    if guess> secret_number:
        print (" Too High ")
    else:
        print(" Too low")
    guess = int (input("Try again: "))
        
print ("Guessed Correctly")
```

### Notice how many times the print statements occurred and how the while loop kept going until the True condition was met, which occurred once x==5. It's important to note that once this occurred the code stopped. Let's see how we could add an else statement:

```python
x = 0

while x < 5:
    print('x is currently: ',x)
    print(' x is still less than 5, adding 1 to x')
    x= x + 1
else:
    print('All Done!')
```

# break, continue, pass

We can use <code>break</code>, <code>continue</code>, and <code>pass</code> statements in our loops to add additional functionality for various cases. The three statements are defined by:

    break: Breaks out of the current closest enclosing loop.
    continue: Goes to the top of the closest enclosing loop.
    pass: Does nothing at all.
    
    
Thinking about <code>break</code> and <code>continue</code> statements, the general format of the <code>while</code> loop looks like this:

    while test: 
        code statement
        if test: 
            break
        if test: 
            continue 
    else:

Let's go ahead and look at some examples!

```python
# Using While Loop
count = 0
while count < 5:
    if count == 3:
        break  # Exit the loop when count is 3
    print(count)
    count = count + 1
```
