# Sets

Sets are an unordered collection of *unique* elements. We can construct them by using the set() function. Let's go ahead and make a set to see how it works.
* <code>No duplicate elements</code>. If try to insert the same item again, it overwrites previous one.
* An <code>unordered collection</code>. When we access all items, they are accessed without any specific order and we cannot access items using <code>indexes</code> as we do in lists.
* Internally use <code>hashing</code> that makes set efficient for search, insert and delete operations. It gives a major advantage over a list for problems with these operations.
* <coode>Mutable</code>, meaning we can add or remove elements after their creation, the individual elements within the set cannot be changed directly.

* No Parameters: Unlike the list.pop() method, set.pop() takes no arguments.
* Arbitrary Selection: It removes a "random" or arbitrary item. The specific element removed is unpredictable because sets  do not maintain order.
* Returns Value: It returns the element that was removed, allowing you to store it in a variable for further use.

#### That's because a set is only concerned with unique elements! We can cast a list with multiple repeat elements to a set to get the unique elements. For example:

**Synatx: Set.update()**  
**Merges another group of values(Iterable) into the set**
* With another set: Adding all items from one set to another.
* With a list: Merging items from a list into a set while removing duplicates.
* With a dictionary: Only the keys of the dictionary are added to the set.
* With a string: The string is treated as an iterable of characters, adding each unique character as an individual element to the set.

```python
cons_set.update("Well")
cons_set
```

```python
cons_set.update([50,100])
cons_set
```

```python
cons_set.update({"apple","mango"})
cons_set
```

**Synatx: Set.union()**  
**union() function combines two sets and returns a new set with all unique elements.**

```python
print(cons_set)
b_set = {1,2,10,20,30}
print(b_set)

combines = cons_set.union(b_set)
print(combines)
```

**Synatx: Set.intersection()**  
**intersection() function returns a new set containing elements that are common to both sets.**

```python
inter_set = cons_set.intersection(b_set)
inter_set
```

**Synatx: Set.defference()**  
**difference() function returns a set containing elements that are in the first set but not in the second.**

```python
# The difference() method keeps the items from the first set that are not in the other set(s).

diff_set = cons_set.difference(b_set)
#diff_set = b_set.difference(cons_set)
diff_set
```

**Get the Length of a Set**
**To determine how many items a set has, use the len() function.**

**Access Items**
**You cannot access items in a set by referring to an index or a key.**

**But you can loop through the set items using a for loop, or ask if a specified value is present in a set, by using the in keyword.**

```python
fruit = {"apple", "banana", "cherry", "mango"}

for x in fruit:
    print(x)
```

```python
print("banana" in fruit)
print("apple" not in fruit)
```
