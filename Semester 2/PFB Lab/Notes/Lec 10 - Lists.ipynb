## The Repetition Operator

### The repetition operator in Python is the asterisk symbol (*). 
### It is used with sequence data types <code>(strings, lists, and tuples)</code> to repeat their elements a specified number of times.
### When you use the <code>* operator</code> with a <code>sequence and an integer n</code>, 
### Python creates a new sequence consisting of the original sequence's elements repeated n times.
* Syntax: sequence * n or n * sequence                                                                                    
* n must be a non-negative integer.                                                                                       
* The original sequence remains unchanged; a new object is returned.

### The range() Function with the for Loop
**The range() function in a Python for loop is used to perform an action a specific number of times or to iterate over a sequence of numbers. It is particularly useful when you need to access the index of elements or simply count iterations.**

**range() function generates the immutable sequence of numbers starting from the given start integer to the stop integer. The range() is a built-in function that returns a range object that consists series of integer numbers, which we can iterate using a for loop.**

**The range() function is memory-efficient because it generates numbers one by one as needed (lazy evaluation), rather than creating an entire list in memory.**

### Syntax and Parameters

The range() function can be called with one, two, or three arguments, all of which must be integers:

* range(stop): Generates a sequence of integers <code>starting from 0 up to, but not including, the stop value</code>, with a default step of 1.
* range(start, stop): Generates a sequence from the <code>start value (inclusive) up to, but not including, the stop value, with a default step of 1</code>.
* range(start, stop, step): Generates a sequence from <code>start to stop, incrementing or decrementing by the step value</code>. The step cannot be zero.

```python
# index start form 0
# holds a series of values that can be iterated over
#         0,1,2,3,4                    iterated over 5 times

# list()
numbers = list(range(5))
```

```python
print(numbers)
```

```python
# range(start, stop, step) stop element is exclosed

numbers = list(range(1, 5, 1))
print(numbers)
```

```python
# range(start, stop, step)

numbers = list(range(0, 11, 2))
print(numbers)
```

**List and String functions are Used in Function**

```python
names = ['aayat','tayyab','kashif']

def capitalized(lists):
    l = []
    for words in lists:
        l.append(words.capitalize())
    return l
print(capitalized(names))    
```

```python
upper = ['KHAN','PAKISTAN']

def is_upper(lists):
    l = []
    for words in lists:
        if words.isupper():
            l.append(words)
    return l
print(is_upper(names))
print(is_upper(upper))
```

```python
upper = ['KHAN','PAKISTAN']

def is_upper(lists):
    l = []
    for words in lists:
        if words[0].isupper():
            l.append(words[0])
    return l
print(is_upper(names))
print(is_upper(upper))
```

# List Comprehensions
Python has an advanced feature called list comprehensions. They allow for quick construction of lists. To fully understand list comprehensions we need to understand for loops.
In addition to sequence operations and list methods, Python includes a more advanced operation called a list comprehension.

List comprehensions allow us to build out lists using a different notation. You can think of it as essentially a one line for loop built **inside of brackets.** For a simple example:

Here are a few examples!

**Basic Syntax:**
The standard structure of a list comprehension is:  
**new_list = [expression for item in iterable if condition]**
* expression: The operation or value to be added to the new list.
* item: The current element being processed from the iterable.
* iterable: The sequence or collection being looped through.
* if condition (optional): A filter that only includes items where the condition is True.

**Example 1**

```python
# Grab every letter in string
for letter in "word":
    print(letter)
```

```python
letter_list = [letter for letter in "word"]
```

```python
letter_list
```

This is the basic idea of a list comprehension. If you're familiar with mathematical notation this format should feel familiar for example: x^2 : x in { 0,1,2...10 }

Let's see a few more examples of list comprehensions in Python:

**Example 2**

```python
# Square numbers in range and turn into list
for num in range(0, 11):
    print(f"{num**2}")
```

```python
squre_numbers = [num**2 for num in range(0, 10+1)]
```

```python
squre_numbers
```

**Example 3**  
Let's see how to add in if statements:

```python
# Check for even numbers in a range (0 to 10)
for num in range(11):
    if num%2 == 0:
        print(num)
```

```python
even_numbers = [num for num in range(11) if num%2==0]
even_numbers
```

**Example 4**  
Can also do more complicated arithmetic:
#### Float formatting follows `"result: {value:{width}.{precision}}"`

```python
# Convert Celsius to Fahrenheit
# (F = 5/9 * (Temp + 32)

celsius = [0,10,20.1,34.5]

for temp in celsius:
    F = ((5/9) * temp + 32)
    print(f'{F:.2f}')
    #print(F)
```

```python
# expression for variable in seq/iter

Fahrenheit = [((5/9) * temp + 32) for temp in celsius]
Fahrenheit
```
